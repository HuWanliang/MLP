```python
from collections import defaultdict

import dnnlpy
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
import torch.utils.data as utils
import torchvision.datasets as datasets
import torchvision.transforms.v2 as v2
from torch import Tensor

torch.manual_seed(42)
    #固定PyTorch全局随机数生成器的种子为42
dnnlpy.set_matplotlib_format('svg')
    #设置Matplotlib输出图像的格式为 SVG（矢量图）
print('PyTorch version:', torch.__version__)
print("MPS available:", torch.backends.mps.is_available())
    #M芯片是否有GPU加速
```

```python
#创建MLP神经网络
class MLP(nn.Module):
    #定义名为MLP的类，继承父类nn.Module（pytorch的神经网络基类）
    def __init__( #定义MLP的初始化函数，用于初始化对象self
            self,#self指的是MLP的对象，model=MLP()，model就是self
            input_dim:int=784,
            hidden_dim:int=256,
            num_classes:int=10,
    ):
        super().__init__()
            #调用父类nn.Module的初始化函数，用于初始化PyTorch管理网络层和参数所需的内部机制
        self.net=nn.Sequential(
            #用torch的容器函数.Sequential()给self对象按顺序组建神经网络层，并赋值到属性net中
            nn.Flatten(),
            nn.Linear(input_dim,hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim,num_classes)
        )

    #forward
    def forward(self,x:Tensor)->Tensor:
        #定义forward成员函数，传入参数为x张量，返回值类型为张量
        return self.net(x)
```

```python
model=MLP()
x=torch.randn(32,1,28,28)
logits=model(x)
print('logits shape:',{logits.shape})
```
